```python
import pandas as pd
import unicodedata

# Update these whenever you re-run the main model notebook
d_m = 480  # Drake — Iceman model prediction
o_m = 414  # Olivia Rodrigo model prediction
a_m = 408  # Ariana Grande model prediction

W = 80

def epad(s, w):
    return s.ljust(w - sum(1 for c in s if unicodedata.east_asian_width(c) in ('W','F')))

def print_tracker(emoji, name, daily_log, weekly_eau, model_pred):
    df_d = pd.DataFrame(daily_log)
    df_w = pd.DataFrame(weekly_eau)
    try:
        parsed = pd.to_datetime(df_d['Date'], format='%Y-%m-%d', errors='coerce')
        df_d['Date'] = parsed.dt.date.astype(str).where(parsed.notna(), df_d['Date'])
    except Exception:
        pass
    week_total = df_d['Streams'].sum()
    days_done  = (df_d['Streams'] > 0).sum()
    any_eau    = (df_w['EAU'] > 0).any()

    if days_done == 0 and not any_eau:
        print(f"  {emoji} {name:<22} no data yet  |  model: {int(model_pred)}M")
        return

    print(f"\n{'='*W}")
    print(f"{f'{emoji}  {name} — TRACKER  (kworb.net | Billboard)':^{W}}")
    print(f"{'='*W}")

    if days_done == 7:
        diff = week_total - model_pred
        print(f"  Wk1 streams: {week_total:.1f}M  |  model: {int(model_pred)}M  |  {diff:+.1f}M ({abs(diff)/model_pred*100:.1f}% {'over' if diff>0 else 'under'})")
    else:
        print(f"  {'DATE':<14}  {'DAILY':>8}  {'CUMUL':>10}  STATUS")
        print(f"  {'-'*50}")
        cum = 0
        for _, r in df_d.iterrows():
            c = r['Streams'] > 0
            if c: cum += r['Streams']
            print(f"  {str(r['Date']):<14}  "
                  f"{f'{r.Streams:.1f}M' if c else '—':>8}  "
                  f"{f'{cum:.1f}M' if c else '—':>10}  "
                  f"{'confirmed' if c else 'pending'}")
        print(f"  {'-'*50}")
        print(f"  {week_total:.1f}M so far ({days_done}/7 days)  |  model: {int(model_pred)}M")

    print(f"\n  {'WEEK':<24}  {'EAU':>10}  {'WoW':>9}  STATUS")
    prev_eau, shown_pending = None, False
    for _, rw in df_w.iterrows():
        ew = rw['EAU']
        if ew > 0:
            wow = f"{(ew-prev_eau)/prev_eau*100:+.1f}%" if prev_eau else "—"
            prev_eau = ew
            print(f"  {rw['Week']:<24}  {ew:>10,}  {wow:>9}  confirmed")
            shown_pending = False
        elif not shown_pending:
            print(f"  {rw['Week']:<24}  {'—':>10}  {'—':>9}  pending")
            shown_pending = True

    confirmed_w = df_w[df_w['EAU'] > 0]
    if not confirmed_w.empty:
        print(f"  Cumulative EAU: {confirmed_w['EAU'].sum():,}  ({len(confirmed_w)}/5 weeks)")
    print(f"{'='*W}")

# =============================================================================
# 🧊  ICEMAN — update Streams daily (kworb.net) | EAU weekly (Billboard)
# =============================================================================
print_tracker('🧊', 'ICEMAN',
    daily_log = {
        'Date':    ['2026-05-15','2026-05-16','2026-05-17','2026-05-18',
                    '2026-05-19','2026-05-20','2026-05-21'],
        'Streams': [140.23, 71.18, 50.81, 54.08, 49.43, 45.37, 44.22]
    },
    weekly_eau = {
        'Week': ['Week 1 (May 15-21)','Week 2 (May 22-28)','Week 3 (May 29-Jun 4)',
                 'Week 4 (Jun 5-11)', 'Week 5 (Jun 12-18)'],
        'EAU':  [463000, 225000, 171000, 134000, 104000]
    },
    model_pred = d_m
)

# =============================================================================
# 💜  OLIVIA RODRIGO — replace TBD dates once album releases
# =============================================================================
print_tracker('💜', 'OLIVIA RODRIGO',
    daily_log = {
        'Date':    ['Day1','Day2','Day3','Day4','Day5','Day6','Day7'],
        'Streams': [81.59, 57.81, 51, 54.95, 52.58, 49.7, 46.97]
    },
    weekly_eau = {
        'Week': ['Week 1','Week 2','Week 3','Week 4','Week 5'],
        'EAU':  [0, 0, 0, 0, 0]
    },
    model_pred = o_m
)

# =============================================================================
# 🌺  ARIANA GRANDE — replace TBD dates once album releases
# =============================================================================
print_tracker('🌺', 'ARIANA GRANDE',
    daily_log = {
        'Date':    ['TBD-Day1','TBD-Day2','TBD-Day3','TBD-Day4',
                    'TBD-Day5','TBD-Day6','TBD-Day7'],
        'Streams': [0, 0, 0, 0, 0, 0, 0]
    },
    weekly_eau = {
        'Week': ['Week 1','Week 2','Week 3','Week 4','Week 5'],
        'EAU':  [0, 0, 0, 0, 0]
    },
    model_pred = a_m
)
```
